```python
from pathlib import Path
import pandas as pd
```

```python
# Carpeta del proyecto de experimentos
BASE = Path.cwd()

CATALOGO = BASE.parent / "00_datos_limpios" / "catalogo_final.csv"

ENTRADA = BASE.parents[1] / "datos/interim_v2"

SALIDA = BASE.parent / "datos_discretizados"

(SALIDA / "interpolacion").mkdir(parents=True, exist_ok=True)
(SALIDA / "percentiles").mkdir(parents=True, exist_ok=True)
```

```python
print(SALIDA)
```

```python
catalogo = pd.read_csv(CATALOGO)

# Seguridad: conservar solo estrellas con 300 o más observaciones
catalogo = catalogo[catalogo["n_obs"] >= 300].copy()

# Construir el mismo identificador que aparece en los archivos discretizados
catalogo["star_id"] = (
    catalogo["proyecto"] + "_" + catalogo["base"]
)

# RR Lyrae
ids_rr = set(
    catalogo.loc[
        catalogo["tipo"] == "RRLYR",
        "star_id"
    ]
)

# Todas las cefeidas
ids_cef = set(
    catalogo.loc[
        catalogo["tipo"].isin(["CEP", "ACEP", "T2CEP"]),
        "star_id"
    ]
)

print(f"RR Lyrae : {len(ids_rr):,}")
print(f"Cefeidas : {len(ids_cef):,}")
print(f"Total     : {len(ids_rr) + len(ids_cef):,}")
```

```python
def filtrar_archivo(archivo_entrada, archivo_rr, archivo_cef):

    print(f"\nProcesando: {archivo_entrada.name}")

    df = pd.read_csv(archivo_entrada)

    rr = df[df["star_id"].isin(ids_rr)]
    cef = df[df["star_id"].isin(ids_cef)]

    rr.to_parquet(archivo_rr, index=False)
    cef.to_parquet(archivo_cef, index=False)

    print(f"RR Lyrae : {len(rr)}")
    print(f"Cefeidas : {len(cef)}")
```

```python
filtrar_archivo(
    ENTRADA / "dataset_50interpolation.csv",
    SALIDA / "interpolacion" / "rrlyrae_50.parquet",
    SALIDA / "interpolacion" / "cefeidas_50.parquet",
)

filtrar_archivo(
    ENTRADA / "dataset_100interpolation.csv",
    SALIDA / "interpolacion" / "rrlyrae_100.parquet",
    SALIDA / "interpolacion" / "cefeidas_100.parquet",
)

filtrar_archivo(
    ENTRADA / "dataset_50percentiles.csv",
    SALIDA / "percentiles" / "rrlyrae_50.parquet",
    SALIDA / "percentiles" / "cefeidas_50.parquet",
)

filtrar_archivo(
    ENTRADA / "dataset_100percentiles.csv",
    SALIDA / "percentiles" / "rrlyrae_100.parquet",
    SALIDA / "percentiles" / "cefeidas_100.parquet",
)

print("\nProceso terminado.")
```
